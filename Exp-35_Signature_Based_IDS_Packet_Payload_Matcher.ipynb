```python
# ==========================================================
# Experiment 7: Signature-Based IDS Packet Payload Matcher
# ==========================================================

# ----------------------------------------------------------
# Signature Database
# ----------------------------------------------------------

SIGNATURES = [

    {
        "sid": 1000001,
        "name": "Possible SQL Injection",
        "pattern": "union select"
    },

    {
        "sid": 1000002,
        "name": "Directory Traversal Attempt",
        "pattern": "../../../etc/passwd"
    },

    {
        "sid": 1000003,
        "name": "Nmap Scripting Engine User-Agent",
        "pattern": "nmap scripting engine"
    }

]


# ----------------------------------------------------------
# Function to scan packet payloads
# ----------------------------------------------------------

def scan_payloads(packets, signatures=SIGNATURES):
    """
    Scan packet payloads for known attack signatures.

    Returns a list of alerts.
    """

    alerts = []

    for index, packet in enumerate(packets):

        payload = packet["payload"].lower()

        for signature in signatures:

            if signature["pattern"] in payload:

                alerts.append({
                    "packet_index": index,
                    "sid": signature["sid"],
                    "name": signature["name"]
                })

    return alerts


# ==========================================================
# Test Cases
# ==========================================================

def test_experiment7():

    packets = [

        {
            "payload": "GET /products?id=1 HTTP/1.1"
        },

        {
            "payload":
            "GET /login?user=admin' UNION SELECT username,password FROM users-- HTTP/1.1"
        },

        {
            "payload":
            "GET /download?file=../../../etc/passwd HTTP/1.1"
        }

    ]

    # Scan payloads
    alerts = scan_payloads(packets)

    # Display Results
    print("IDS Alert Report")
    print("-" * 60)

    if alerts:

        for alert in alerts:

            print(f"Packet Index : {alert['packet_index']}")
            print(f"Signature ID : {alert['sid']}")
            print(f"Alert Name   : {alert['name']}")
            print()

    else:
        print("No malicious payloads detected.")

    # Assertions
    alert_indices = {alert["packet_index"] for alert in alerts}

    assert 0 not in alert_indices
    assert 1 in alert_indices
    assert 2 in alert_indices

    assert any(
        alert["name"] == "Possible SQL Injection"
        for alert in alerts
    )

    assert any(
        alert["name"] == "Directory Traversal Attempt"
        for alert in alerts
    )

    print("All test cases passed.")


# ----------------------------------------------------------
# Run Test
# ----------------------------------------------------------

test_experiment7()
```
