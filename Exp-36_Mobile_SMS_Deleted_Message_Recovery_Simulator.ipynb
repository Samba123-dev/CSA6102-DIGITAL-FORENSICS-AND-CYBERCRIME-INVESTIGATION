```python
# ==========================================================
# Experiment 8: Mobile SMS Deleted-Message Recovery Simulator
# ==========================================================

# ----------------------------------------------------------
# Function to recover deleted SMS messages
# ----------------------------------------------------------

def recover_deleted_messages(sms_table):
    """
    Return SMS messages that are marked as deleted
    but have not yet been overwritten.
    """

    recoverable = [
        row
        for row in sms_table
        if row["is_deleted"] and not row["overwritten"]
    ]

    return recoverable


# ----------------------------------------------------------
# Function to summarize SMS database
# ----------------------------------------------------------

def summarize_table(sms_table):
    """
    Count active, recoverable and permanently lost messages.
    """

    active = [
        row
        for row in sms_table
        if not row["is_deleted"]
    ]

    recoverable = recover_deleted_messages(sms_table)

    lost = [
        row
        for row in sms_table
        if row["is_deleted"] and row["overwritten"]
    ]

    return {
        "active": len(active),
        "recoverable": len(recoverable),
        "permanently_lost": len(lost)
    }


# ==========================================================
# Test Cases
# ==========================================================

def test_experiment8():

    sms_table = [

        {
            "rowid": 1,
            "address": "+1-555-0101",
            "body": "See you at 6pm",
            "date": "2026-01-01",
            "is_deleted": False,
            "overwritten": False
        },

        {
            "rowid": 2,
            "address": "+1-555-0199",
            "body": "Transfer the funds now, delete after reading",
            "date": "2026-01-02",
            "is_deleted": True,
            "overwritten": False
        },

        {
            "rowid": 3,
            "address": "+1-555-0150",
            "body": "Old spam message",
            "date": "2025-06-01",
            "is_deleted": True,
            "overwritten": True
        }

    ]

    # Recover deleted messages
    recoverable = recover_deleted_messages(sms_table)

    print("Recoverable Deleted Messages")
    print("-" * 60)

    for message in recoverable:
        print(f"Row ID    : {message['rowid']}")
        print(f"Address   : {message['address']}")
        print(f"Body      : {message['body']}")
        print(f"Date      : {message['date']}")
        print()

    # Summary
    summary = summarize_table(sms_table)

    print("Database Summary")
    print("-" * 60)

    print(f"Active Messages          : {summary['active']}")
    print(f"Recoverable Messages     : {summary['recoverable']}")
    print(f"Permanently Lost         : {summary['permanently_lost']}")
    print()

    # Assertions
    assert len(recoverable) == 1
    assert recoverable[0]["rowid"] == 2
    assert "Transfer the funds" in recoverable[0]["body"]

    assert summary == {
        "active": 1,
        "recoverable": 1,
        "permanently_lost": 1
    }

    print("All test cases passed.")


# ----------------------------------------------------------
# Run Test
# ----------------------------------------------------------

test_experiment8()
```
