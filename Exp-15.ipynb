```
import pandas as pd
import hashlib
from collections import Counter

# ==========================
# Load Dataset
# ==========================
df = pd.read_csv("Malware_Hash_Database (1).csv")

# ==========================
# Generate Hashes
# ==========================
def md5_hash(text):
    return hashlib.md5(str(text).encode()).hexdigest()

def sha1_hash(text):
    return hashlib.sha1(str(text).encode()).hexdigest()

def sha256_hash(text):
    return hashlib.sha256(str(text).encode()).hexdigest()

df["MD5"] = df["Content"].apply(md5_hash)
df["SHA1"] = df["Content"].apply(sha1_hash)
df["SHA256"] = df["Content"].apply(sha256_hash)

# ==========================
# Sample Malware Database
# ==========================
known_malware = [
    "invoice_update.exe",
    "payroll_viewer.scr",
    "remote_access.exe",
    "key_input.exe"
]

df["Known_Malware"] = df["File_Name"].isin(known_malware)

# ==========================
# Hash-Based Detection
# ==========================
malware_hashes = set(df[df["Known_Malware"]]["SHA256"])

df["Detection"] = df["SHA256"].apply(
    lambda x: "Malware" if x in malware_hashes else "Safe"
)

# ==========================
# Hash Family Clustering
# ==========================
def family(file):
    file = str(file).lower()
    if "trojan" in file:
        return "Trojan"
    elif "ransom" in file:
        return "Ransomware"
    elif "key" in file:
        return "Keylogger"
    elif "bank" in file:
        return "Banking Malware"
    elif "remote" in file:
        return "Backdoor"
    else:
        return "Benign"

df["Family"] = df["File_Name"].apply(family)

# ==========================
# Print Results
# ==========================
print(df[["File_Name", "MD5", "SHA1", "SHA256", "Detection", "Family"]])

print("\nDetection Statistics")
print(df["Detection"].value_counts())

# Save output
df.to_csv("Malware_Hash_Database_Output.csv", index=False)

print("\nOutput saved as Malware_Hash_Database_Output.csv")
```
