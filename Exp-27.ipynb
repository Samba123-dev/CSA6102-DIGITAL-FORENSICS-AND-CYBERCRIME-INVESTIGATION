```
# ==========================================================
# Experiment 10: Slack Space Calculation
# Aim:
# Calculate slack space – the unused bytes left over in the
# last disk cluster allocated to a file.
# ==========================================================

# Typical disk cluster size (in bytes)
CLUSTER_SIZE = 4096

# ----------------------------------------------------------
# Function to Calculate Slack Space
# ----------------------------------------------------------
def calculate_slack_space(file_size, cluster_size=CLUSTER_SIZE):
    # Ceiling division to determine required clusters
    clusters_needed = -(-file_size // cluster_size)

    # Total allocated space
    allocated_space = clusters_needed * cluster_size

    # Unused bytes in the last cluster
    slack_space = allocated_space - file_size

    return allocated_space, slack_space

# ----------------------------------------------------------
# Example File Size
# ----------------------------------------------------------
file_size = 5000  # bytes

allocated, slack = calculate_slack_space(file_size)

# ----------------------------------------------------------
# Display Output
# ----------------------------------------------------------
print(f"File size: {file_size} bytes -> allocated: {allocated} bytes, slack space: {slack} bytes")

# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------
def test_experiment10():

    assert allocated == 8192, \
        "5000 bytes needs 2 clusters of 4096 bytes = 8192 bytes allocated"

    assert slack == 8192 - 5000, \
        "Slack space should be the unused remainder of the last cluster"

    allocated2, slack2 = calculate_slack_space(4096)

    assert slack2 == 0, \
        "A file that exactly fills a cluster should have zero slack space"

    print("Experiment 10: All test cases passed.")

# Run Test Cases
test_experiment10()
```
